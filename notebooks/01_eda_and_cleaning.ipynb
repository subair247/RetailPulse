## Day 1: Exploratory Data Analysis & Data Validation

```python
# Import all packages
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

1. load clean datasets

```python
print("Loading datasets from data/clean/...")
customers_df = pd.read_csv('../data/clean/customer_master.csv')
inventory_df = pd.read_csv('../data/clean/inventory_snapshot.csv')
promotions_df = pd.read_csv('../data/clean/promotions.csv')
sku_df = pd.read_csv('../data/clean/sku_inventory_flags.csv')
sku_master_df = pd.read_csv('../data/clean/sku_master.csv')
store_df = pd.read_csv('../data/clean/store_master.csv')

print("\n--- Datasets Shapes ---")
print(f"Customers: {customers_df.shape}")
print(f"inventory: {inventory_df.shape}")
print(f"promotions: {promotions_df.shape}")
print(f"sku_inventory: {sku_df.shape}")
print(f"sku_masters: {sku_master_df.shape}")
print(f"store_master: {store_df.shape}")
```

2. Loading Sales Transactions in Chunks (Since it's ~10M rows / 784MB)

```python
print("\n--- Loading Sales Transactions (chunking method) ---")
chunk_size = 1000000
sales_sample = None 

try:
    for chunk in pd.read_csv('../data/clean/sales_transactions.csv', chunksize=chunk_size):
        sales_sample = chunk.head(1000)
        print(f"Loaded a sales chunk. Total Columns: {chunk.shape[1]}")
        break
except FileNotFoundError:
    print("⚠️ sales_transactions.csv is not uploaded yet. Skipping chunk loading safely.")

print("Sales Sample Columns:", sales_sample.columns.tolist() if sales_sample is not None else "File not found / Not uploaded")
```

3. Basic Missing Values Check

```python
print("\n--- Missing Values Check (Master Tables) ---")

print("Customers Nulls:", customers_df.isnull().sum().sum())
print("SKU Master Nulls:", sku_master_df.isnull().sum().sum())
print("Store Master Nulls:", store_df.isnull().sum().sum())

print("\nDay 1 Dataset Loading & Verification Completed Successfully!")
```
