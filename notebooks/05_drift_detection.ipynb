### Day 3: Data Drift Monitoring & Validation

```python
import pandas as pd
import numpy as np 
import os 
print("Initializing Data Drift & Model Monitoring Pipeline...")
```

1. Load Reference (Training) and Current (Production) Datasets

```python
from numpy import size
sku_path = '../data/clean/sku_master.csv' if os.path.exists('../data/clean/sku_master.csv') else 'sku_master.csv'
df_ref = pd.read_csv(sku_path)
# Simulate current production data sample
df_current = df_ref.sample(frac=0.8, random_state=42).copy()
# Add slight synthetic drift to current pricing/costs
df_current['unit_price'] = df_current['unit_price'] * np.random.normal(1.05, 0.05, size=len(df_current))

print(f"Reference Dataset Shape: {df_ref.shape}")
print(f"Current Production Dataset Shape: {df_current.shape}")
```

2. Simple Statistical Drift Calculation (Mean Price Comparison)

```python
ref_mean_price = df_ref['unit_price'].mean()
curr_mean_price = df_current['unit_price'].mean()

price_shift_pct = ((curr_mean_price - ref_mean_price) / ref_mean_price) * 100
drift_detected = abs(price_shift_pct) > 3.0

print(f"\n--- Drift Analysis Results ---")
print(f"Reference Mean Unit Price: ${ref_mean_price:.2f}")
print(f"Current Production Mean Unit Price: ${curr_mean_price:.2f}")
print(f"Percentage Shift: {price_shift_pct:.2f}%")
print(f"Data Drift Alert Status: {'⚠️ DRIFT DETECTED (Retraining Recommended)' if bool(drift_detected) else '✅ STABLE (No Significant Drift)'}")
```

3. Export Drift Monitoring Report

```python
os.makedirs('../reports', exist_ok=True)
drift_report = pd.DataFrame({
    'Metric': ['Reference Mean Price', 'Current Mean Price', 'Percentage Shift (%)', 'Drift Status'],
    'Value': [ref_mean_price, curr_mean_price, price_shift_pct, 'Drift Detected' if drift_detected else 'Stable']
})
drift_report.to_csv('../reports/drift_detection_report.csv', index=False)

print("\nData Drift Report saved successfully to ../reports/")
print("\nRetailPulse Project Pipeline Completed Successfully from End-to-End!")
```
